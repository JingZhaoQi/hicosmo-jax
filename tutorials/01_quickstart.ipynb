# HIcosmo 快速入门

本教程展示 HIcosmo 的核心功能，5 分钟上手宇宙学参数推断。

**核心 API**：
- `hicosmo()` — 3 行代码完成参数推断
- `Plotter` — 结果可视化

````python
import hicosmo as hc
hc.init()
````

## 1. 极简 API（3 行代码）

````python
from hicosmo import hicosmo

# 3 行代码：LCDM + BAO 数据
mcmc = hicosmo('LCDM', 'bao', ['H0', 'Omega_m'], chain_name='quickstart_bao')
mcmc.run(num_samples=2000)
mcmc.print_summary()
````

````python
# 可视化
from hicosmo.visualization import Plotter

plotter = Plotter(mcmc)  # 直接传 MCMC 对象
plotter.corner(['H0', 'Omega_m'], filename='figures/01_bao_corner.pdf')
````

## 2. 多数据联合分析

````python
# 联合分析：BAO + CMB
mcmc_joint = hicosmo('LCDM', ['bao', 'planck'], ['H0', 'Omega_m'], chain_name='quickstart_joint')
mcmc_joint.run(num_samples=2000)
mcmc_joint.print_summary()
````

````python
Plotter(mcmc_joint).corner(['H0', 'Omega_m'], filename='figures/01_joint_corner.pdf')
````

## 3. 多链比较

````python
# 比较 BAO-only vs BAO+CMB
plotter_cmp = Plotter(
    [mcmc.chain_name, mcmc_joint.chain_name],
    labels=['BAO', 'BAO+CMB']
)
plotter_cmp.corner(['H0', 'Omega_m'], filename='figures/01_comparison.pdf')
````

## 4. 统计报告

````python
Plotter(mcmc_joint).report()
````
